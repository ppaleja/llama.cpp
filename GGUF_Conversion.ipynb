```python
# Download the model from Hugging Face
from huggingface_hub import snapshot_download

model_id = "Motif-Technologies/Motif-2.6b-v1.1-LC"
print(f"Downloading {model_id}...")
snapshot_download(repo_id=model_id, local_dir=model_id, local_dir_use_symlinks=False)
print("Download complete.")
```

```python
# Convert the model to GGUF
print(f"Converting {model_id} to GGUF...")
get_ipython().system(f'python3 convert_hf_to_gguf.py {model_id} --outtype f16 --verbose')
```
