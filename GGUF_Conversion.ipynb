```python
import gguf
import sys
import numpy as np
print(f"Verifying {gguf_path}...")
try:
    reader = gguf.GGUFReader(gguf_path)
except Exception as e:
    print(f"Error reading GGUF file: {e}")
```

```python
print("\n--- Key-Value Pairs (Motif/Attention specific) ---")
# assert architecture is motif
arch_key = next((k for k in reader.fields.keys() if "general.architecture" in k), None)
if arch_key:
    arch = str(reader.fields[arch_key].parts[-1], encoding='utf-8')
    print(f"Architecture: {arch}")
    assert arch == "motif", f"Expected architecture 'motif', got '{arch}'"
else:
    print("WARNING: Could not find general.architecture key")
```

```python
relevant_keys = ["grouped_ratio", "k_ratio", "lambda_init", "num_noise_heads"]
found_keys = {}
for key, field in reader.fields.items():
    if any(k in key for k in relevant_keys):
          val = field.parts[-1]
          if isinstance(val, list): val = val[0] # Unpack single-item lists
          found_keys[key.split('.')[-1]] = val
          print(f"  {key}: {val}")
```

```python
# Assertions for keys
if "grouped_ratio" in found_keys:
    # We expect 4.0 for Motif-2-12.7B
    assert found_keys["grouped_ratio"] == 4.0, f"Expected grouped_ratio=4.0, got {found_keys['grouped_ratio']}"
```

```python
print("\n--- Tensors Check ---")
tensors = reader.tensors
print(f"  Total tensors found: {len(tensors)}")
```

This section verifies that the converted GGUF can be loaded by llama.cpp (hyperparameters parsed and tensors recognized) without building any inference graphs.

Success criteria:
- `llama-cli --dry-run` exits with code 0
- No warnings like "skipping unknown tensor" or "unknown tensor type"
- Architecture matches `motif` (checked earlier)

```python
import subprocess, shlex, re

print("Running llama-cli with minimal inference to verify tensor loading...")
# Use -n 1 (generate only 1 token) and -e (exit after generation) to quickly test loading
cmd = f"/content/llama.cpp/build/bin/llama-cli -m {gguf_path} -p 'test' -n 1 -e 2>&1"
res = subprocess.run(cmd, shell=True, capture_output=True, text=True)

print("Exit code:", res.returncode)
output = res.stdout + res.stderr
print("\n--- Output (first 2000 chars) ---\n", output[:2000])

# Check for loading errors and success indicators
patterns = [
    (r"skipping unknown tensor", False),
    (r"unknown tensor type", False),
    (r"failed", False),
    (r"error reading", False),
    (r"exception", False),
    (r"loaded.*model", True),   # Success indicator
    (r"prompt", True),  # Means it parsed input
]

issues = []
success_found = False
for pattern, is_success in patterns:
    if re.search(pattern, output, re.IGNORECASE):
        if is_success:
            success_found = True
        else:
            issues.append(pattern)

if res.returncode != 0 and not success_found:
    # Check if it's just a different exit code despite successful load
    if "loaded" not in output.lower():
        issues.append(f"Process exited with code {res.returncode}")

if issues:
    print(f"\n[WARNING] Potential loading issues detected: {issues}")
    print("Full output:")
    print(output)
else:
    print("\n✓ PASS: Model loaded and began processing without unknown/skipped tensor errors.")
```

# Phase 4: Graph Building Verification

This section verifies that the inference graph can be built and executed successfully. This tests the actual graph construction logic in `llm_build_motif`.

Success criteria:
- `llama-cli` can build the inference graph without errors
- Graph executes successfully and generates tokens
- No graph-related errors (e.g., "failed to build graph", "unknown operation", shape mismatches)
- Output is reasonable (non-empty, no NaN/inf values)


```python
import subprocess
import shlex
import re

print("Running llama-cli inference to verify graph building...")
print("This will build the inference graph and generate a few tokens.\n")

# Run inference with a simple prompt, generating 5 tokens
# Use deterministic settings for reproducibility
cmd = f"/content/llama.cpp/build/bin/llama-cli -m {gguf_path} -p 'Hello, my name is' -n 5 --temp 0 --top-k 1 --top-p 1 --seed 42"
res = subprocess.run(shlex.split(cmd), capture_output=True, text=True, timeout=300)

print("Exit code:", res.returncode)
print("\n--- stdout (first 2000 chars) ---\n", res.stdout[:2000])
print("\n--- stderr (first 2000 chars) ---\n", res.stderr[:2000])

# Check for graph-related errors
combined = (res.stdout or "") + "\n" + (res.stderr or "")
graph_error_patterns = [
    r"failed to build",
    r"graph.*error",
    r"unknown operation",
    r"shape mismatch",
    r"dimension.*mismatch",
    r"tensor.*not found",
    r"build.*failed",
    r"assertion.*failed",
    r"segmentation fault",
    r"abort",
    r"fatal"
]

graph_issues = []
for pattern in graph_error_patterns:
    matches = re.findall(pattern, combined, re.IGNORECASE)
    if matches:
        graph_issues.extend(matches)

# Check if we got actual output (not just errors)
has_output = bool(re.search(r'[a-zA-Z0-9]{3,}', res.stdout))

# Verify exit code
assert res.returncode == 0, f"llama-cli exited with {res.returncode}"
assert not graph_issues, f"Detected graph building/execution issues: {graph_issues}"
assert has_output, "No output generated - graph may not have executed properly"

print("\n✅ PASS: Graph built and executed successfully!")
print(f"   Generated output: {res.stdout[:100]}...")
```

```python
# Additional verification: Check for Motif-specific operations in verbose output
print("Running verbose inference to verify Motif-specific graph operations...\n")

# Run with verbose logging to see graph operations
cmd_verbose = f"/content/llama.cpp/build/bin/llama-cli -m {gguf_path} -p 'Test' -n 1 --temp 0 --top-k 1 --log-verbose 2>&1 | head -100"
res_verbose = subprocess.run(cmd_verbose, shell=True, capture_output=True, text=True, timeout=60)

print("--- Verbose output (first 3000 chars) ---\n", res_verbose.stdout[:3000])

# Check for Motif-specific operations/patterns
motif_patterns = [
    r"attn_lambda",
    r"polynorm",
    r"grouped.*diff",
    r"attn_sub_norm",
    r"build_grouped",
    r"build_polynorm"
]

found_motif_ops = []
for pattern in motif_patterns:
    matches = re.findall(pattern, res_verbose.stdout, re.IGNORECASE)
    if matches:
        found_motif_ops.extend(set(matches))  # Remove duplicates

if found_motif_ops:
    print(f"\n✅ Found Motif-specific operations: {set(found_motif_ops)}")
else:
    print("\n⚠️  Note: Motif-specific operation names not found in verbose output")
    print("   (This may be normal if operations are named generically)")

# Check for any errors in verbose output
verbose_errors = re.findall(r'error|failed|exception', res_verbose.stdout, re.IGNORECASE)
if verbose_errors:
    print(f"\n⚠️  Warnings/errors in verbose output: {set(verbose_errors)}")
else:
    print("\n✅ No errors detected in verbose output")
```

```python
# Simple HuggingFace test script
# NOTE: If this cell fails with numpy errors, please RESTART THE RUNTIME (Runtime -> Restart session)
# and run this cell again.
import sys
import subprocess

# Install dependencies for quantization
subprocess.check_call([sys.executable, "-m", "pip", "install", "numpy<2", "bitsandbytes", "accelerate"])

try:
    from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
    import torch
    import numpy as np
except ImportError as e:
    print("\n" + "="*60)
    print("ERROR: Import failed. Please RESTART THE RUNTIME (Runtime > Restart session).")
    print("="*60 + "\n")
    raise e

print(f"Numpy version: {np.__version__}")

model_path = "/content/Motif-2-12.7B-Instruct"
print(f"Loading model from {model_path} in 4-bit to fit on T4 GPU...")

try:
    tokenizer = AutoTokenizer.from_pretrained(model_path, trust_remote_code=True)

    # Configure 4-bit quantization
    quantization_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_compute_dtype=torch.bfloat16
    )

    # Load model with quantization
    # device_map="auto" handles moving layers to GPU automatically
    model = AutoModelForCausalLM.from_pretrained(
        model_path,
        trust_remote_code=True,
        quantization_config=quantization_config,
        device_map="auto"
    )

    # Test same prompts
    for prompt in ["goodbye", "hello"]:
        print(f"\nGenerating for: '{prompt}'")
        inputs = tokenizer(prompt, return_tensors="pt").to('cuda')
        outputs = model.generate(**inputs, max_new_tokens=32, do_sample=False)
        print(f"Result: {tokenizer.decode(outputs[0])}")

except Exception as e:
    print(f"An error occurred during model loading/generation: {e}")
```
